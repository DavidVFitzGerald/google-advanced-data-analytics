### Create the training and testing data

Put 70% of the data into a training set and the remaining 30% into a testing set. Create an X and y DataFrame with only the necessary variables.

Build a logistic regression model and fit the model to the training data.

### Create a plot of your model

Create a plot of your model to visualize results using the seaborn package.

```python
sns.regplot(data=df_subset, x="Inflight entertainment", y="satisfaction", logistic=True, ci=None)
```

```python
import numpy as np
import matplotlib.pyplot as plt

X_test_model = X_test.to_numpy()
y_pred_proba = clf.predict_proba(X_test_model)[:,-1]
y_pred = clf.predict(X_test_model)

# Plot the actual data points as a scatter plot
plt.scatter(X_test, y_test, color='black', label="Test data")

# Plot the regression line
# Sort the X values for a continuous line in case of non-linear x data
sort_idx = np.argsort(X_test_model.flatten())
plt.plot(X_test_model[sort_idx], y_pred_proba[sort_idx], color="blue", linewidth=2, label="Regression line probabilities")
plt.plot(X_test_model[sort_idx], y_pred[sort_idx], color="green", linewidth=2, label="Regression line predictions")

plt.xlabel("Inflight entertainment")
plt.ylabel("satisfaction")
plt.title("Logistic Regression Plot")
plt.legend()
plt.show()
```

### Analyze the results

Print out the model's accuracy, precision, recall, and F1 score.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print(f"Accuracy: {accuracy_score(y_test, y_pred):.3f}")
print(f"Precision: {precision_score(y_test, y_pred):.3f}")
print(f"Recall: {recall_score(y_test, y_pred):.3f}")
print(f"F1: {f1_score(y_test, y_pred):.3f}")
```

### Produce a confusion matrix

Data professionals often like to know the types of errors made by an algorithm. To obtain this information, produce a confusion matrix.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred, labels = clf.classes_)
cm
```

```python
np.round(cm/sum(cm), 2)
```

```python
ConfusionMatrixDisplay(confusion_matrix = cm,display_labels = clf.classes_).plot()
```

**Question:** What stands out to you about the confusion matrix?

There are around 14000 true negatives and 17000 true positives. There are 3900 false positives and 3800 false negatives.

The number of correctly estimated values are an order of magnitude higher than the number of wrongly estimated values.
